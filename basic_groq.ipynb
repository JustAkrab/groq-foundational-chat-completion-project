Remember to register your api, use this documentation: https://console.groq.com/docs/overview

```python
from groq import Groq
from dotenv import load_dotenv
import os

load_dotenv()

client = Groq(api_key= os.environ.get("GROQ_API_KEY"),)
```

OK now that you've registered your api_key, try to make a query

```python
chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "Explan your capabilities, and limitations",
        }

    ],
    model="openai/gpt-oss-20b",
)

print(chat_completion.choices[0].message.content)
```

Great, you've done a basic query, now let's see if we can make a working chatbot from this by streaming it. Follow this guide: https://console.groq.com/docs/text-chat

Before we do that, let's design the behavior of the chat bot. So how we do this will be posted in the code below: 

```python
chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant for job hunting",
            # This returns a humane response to the user
            "role": "user",
            "content": "What is your role as a system?",
        }

    ],
    model="openai/gpt-oss-20b",
)

print(chat_completion.choices[0].message.content)
```

Ok now we add streaming:

```python
stream = client.chat.completions.create(
    messages=[
        #explain it's role:
        {"role": "system",
        "content": "You are a job-hunting guide for job seekers. All other topics are irrelevant, do not discuss capabilities outside of job hunting",},

        # Message for assistant to respond to:
        {"role": "user",
        "content": "Explain your capabilities in job hunting briefly",},
    ],


    model="openai/gpt-oss-20b",
    # optional parameters included:

    # temperature is the parameter that controls randomness, the lower the number, the less random.
    temperature = 0.3,

    # The maximum number of tokens to generate. Requests can use up to
    # 2048 tokens shared between prompt and completion.
    max_completion_tokens=1024,


     # Controls diversity via nucleus sampling: 0.5 means half of all
    # likelihood-weighted options are considered.
    top_p=1,


    # A stop sequence is a predefined or user-specified text string that
    # signals an AI to stop generating content, ensuring its responses
    # remain focused and concise. Examples include punctuation marks and
    # markers like "[end]".
    stop=None,


    # If set, partial message deltas will be sent.
    stream=True,
)


for chunk in stream:
    print(chunk.choices[0].delta.content or "", end="")
```
